```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define file paths and corresponding years
file_paths = {
    2014: "/mnt/data/RV_M_M_2014.xlsx",
    2015: "/mnt/data/RV_M_M_2015.xlsx",
    2016: "/mnt/data/RV_M_M_2016.xlsx",
    2017: "/mnt/data/RV_M_M_2017.xlsx",
    2018: "/mnt/data/RV_M_M_2018.xlsx",
    2019: "/mnt/data/RV_M_M_2019.xlsx",
    2020: "/mnt/data/RV_M_M_2020.xlsx",
    2021: "/mnt/data/RV_M_M_2021.xlsx",
    2022: "/mnt/data/RV_M_M_2022.xlsx",
    2023: "/mnt/data/RV_M_M_2023.xlsx",
    2024: "/mnt/data/RV_M_M_2024.xlsx",
    2025: "/mnt/data/RV_M_M_2025.xlsx"
}

# Define the MAKE and MODEL to track
target_make = "ISUZU"
target_model = "HARDBODY NP300 2.5 TDi HiRider"

# Placeholder for combined data
combined_data = []

# Load, filter, and tag each dataset
for year, path in file_paths.items():
    try:
        df = pd.read_excel(path)
        # Normalize column names just in case
        df.columns = df.columns.str.strip().str.upper()
        # Filter for target MAKE and MODEL
        filtered = df[
            (df["MAKE"].str.upper() == target_make.upper()) &
            (df["MODEL"].str.upper() == target_model.upper())
        ].copy()
        filtered["SOURCE_YEAR"] = year
        combined_data.append(filtered)
    except Exception as e:
        print(f"Failed to load or process {year}: {e}")

# Concatenate all years' filtered data
rv_data = pd.concat(combined_data, ignore_index=True)

# Display sample of final combined data
import ace_tools as tools; tools.display_dataframe_to_user(name="Filtered RV Data (ISUZU HARDBODY)", dataframe=rv_data)

# Plot RV% over Age for each Source_Year
plt.figure(figsize=(12, 6))
sns.lineplot(data=rv_data, x="AGE", y="RV%", hue="SOURCE_YEAR", marker="o", palette="tab10")
plt.title("RV% vs Age for ISUZU HARDBODY NP300 2.5 TDi HiRider (2014–2025)")
plt.xlabel("Age")
plt.ylabel("Residual Value (RV%)")
plt.legend(title="Year")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Reload and re-filter by checking if MODEL contains the search string without using MAKE
search_model = "HARDBODY NP300 2.5 TDi HiRider"

# Reload data using only MODEL matching (case-insensitive containment)
combined_data_v2 = []

for year, path in file_paths.items():
    try:
        df = pd.read_excel(path)
        df.columns = df.columns.str.strip().str.upper()
        # Use contains instead of exact match
        filtered = df[df["MODEL"].str.upper().str.contains(search_model.upper(), na=False)].copy()
        filtered["SOURCE_YEAR"] = year
        combined_data_v2.append(filtered)
    except Exception as e:
        print(f"Error loading {year}: {e}")

# Concatenate all valid filtered data
rv_data_v2 = pd.concat(combined_data_v2, ignore_index=True)

# Display the new combined result for visual inspection
import ace_tools as tools; tools.display_dataframe_to_user(name="Filtered RV Data by Model Only", dataframe=rv_data_v2)
```

```python
# Now plot the correct data: RV% vs Age over Source_Year for the filtered HARDBODY model
plt.figure(figsize=(12, 6))
sns.lineplot(data=rv_data_v2, x="AGE", y="RV%", hue="SOURCE_YEAR", marker="o", palette="Spectral")
plt.title("RV% vs Age for NISSAN HARDBODY NP300 2.5 TDi HiRider (2014–2025)")
plt.xlabel("Vehicle Age (Years)")
plt.ylabel("Residual Value (%)")
plt.legend(title="Dataset Year", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
