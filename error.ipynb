```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define file paths and corresponding years
file_paths = {
    2014: "/mnt/data/RV_M_M_2014.xlsx",
    2015: "/mnt/data/RV_M_M_2015.xlsx",
    2016: "/mnt/data/RV_M_M_2016.xlsx",
    2017: "/mnt/data/RV_M_M_2017.xlsx",
    2018: "/mnt/data/RV_M_M_2018.xlsx",
    2019: "/mnt/data/RV_M_M_2019.xlsx",
    2020: "/mnt/data/RV_M_M_2020.xlsx",
    2021: "/mnt/data/RV_M_M_2021.xlsx",
    2022: "/mnt/data/RV_M_M_2022.xlsx",
    2023: "/mnt/data/RV_M_M_2023.xlsx",
    2024: "/mnt/data/RV_M_M_2024.xlsx",
    2025: "/mnt/data/RV_M_M_2025.xlsx"
}

# Define the MAKE and MODEL to track
target_make = "ISUZU"
target_model = "HARDBODY NP300 2.5 TDi HiRider"

# Placeholder for combined data
combined_data = []

# Load, filter, and tag each dataset
for year, path in file_paths.items():
    try:
        df = pd.read_excel(path)
        # Normalize column names just in case
        df.columns = df.columns.str.strip().str.upper()
        # Filter for target MAKE and MODEL
        filtered = df[
            (df["MAKE"].str.upper() == target_make.upper()) &
            (df["MODEL"].str.upper() == target_model.upper())
        ].copy()
        filtered["SOURCE_YEAR"] = year
        combined_data.append(filtered)
    except Exception as e:
        print(f"Failed to load or process {year}: {e}")

# Concatenate all years' filtered data
rv_data = pd.concat(combined_data, ignore_index=True)

# Display sample of final combined data
import ace_tools as tools; tools.display_dataframe_to_user(name="Filtered RV Data (ISUZU HARDBODY)", dataframe=rv_data)

# Plot RV% over Age for each Source_Year
plt.figure(figsize=(12, 6))
sns.lineplot(data=rv_data, x="AGE", y="RV%", hue="SOURCE_YEAR", marker="o", palette="tab10")
plt.title("RV% vs Age for ISUZU HARDBODY NP300 2.5 TDi HiRider (2014–2025)")
plt.xlabel("Age")
plt.ylabel("Residual Value (RV%)")
plt.legend(title="Year")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Re-import needed libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define re-uploaded file paths
file_paths = {
    2014: "/mnt/data/RV_M_M_2014.xlsx",
    2015: "/mnt/data/RV_M_M_2015.xlsx",
    2016: "/mnt/data/RV_M_M_2016.xlsx",
    2017: "/mnt/data/RV_M_M_2017.xlsx",
    2018: "/mnt/data/RV_M_M_2018.xlsx",
    2019: "/mnt/data/RV_M_M_2019.xlsx",
    2020: "/mnt/data/RV_M_M_2020.xlsx",
    2021: "/mnt/data/RV_M_M_2021.xlsx",
    2022: "/mnt/data/RV_M_M_2022.xlsx",
    2023: "/mnt/data/RV_M_M_2023.xlsx"
}

# Load and filter models containing "NP300" across all years
np300_combined_data = []

for year, path in file_paths.items():
    try:
        df = pd.read_excel(path)
        df.columns = df.columns.str.strip().str.upper()
        filtered = df[df["MODEL"].str.upper().str.contains("NP300", na=False)].copy()
        filtered["SOURCE_YEAR"] = year
        np300_combined_data.append(filtered)
    except Exception as e:
        print(f"Error loading {year}: {e}")

# Combine all NP300 data
np300_data = pd.concat(np300_combined_data, ignore_index=True)

# Classify into broader variant categories
def classify_variant(model):
    model_upper = model.upper()
    if "HIRIDER" in model_upper:
        return "HiRider"
    elif "BASE" in model_upper:
        return "Base"
    elif "LE" in model_upper:
        return "LE"
    elif "SE" in model_upper:
        return "SE"
    elif "XE" in model_upper:
        return "XE"
    elif "4X4" in model_upper:
        return "4x4"
    elif "S/C" in model_upper or "SINGLE CAB" in model_upper:
        return "Single Cab"
    elif "D/C" in model_upper or "DOUBLE CAB" in model_upper:
        return "Double Cab"
    else:
        return "Other"

np300_data["VARIANT"] = np300_data["MODEL"].apply(classify_variant)

# Display a sample of how variants were classified
import ace_tools as tools; tools.display_dataframe_to_user(name="NP300 Variant Classification", dataframe=np300_data[["MODEL", "VARIANT", "AGE", "RV%", "SOURCE_YEAR"]].head(20))
```

```python
# Filter out rows with missing or zero RV% for better plotting
np300_cleaned = np300_data.dropna(subset=["RV%", "AGE"])
np300_cleaned = np300_cleaned[np300_cleaned["RV%"] > 0]

# Plot RV% vs Age for each NP300 variant over time
plt.figure(figsize=(14, 7))
sns.lineplot(data=np300_cleaned, x="AGE", y="RV%", hue="VARIANT", style="SOURCE_YEAR", markers=True)
plt.title("RV% vs Age by NP300 Variant (2014–2023)")
plt.xlabel("Vehicle Age")
plt.ylabel("Residual Value (%)")
plt.legend(title="Variant / Year", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Create a facet plot: one chart per NP300 variant, RV% vs Age, over dataset years
g = sns.FacetGrid(np300_cleaned, col="VARIANT", col_wrap=3, height=4, sharey=True)
g.map_dataframe(sns.lineplot, x="AGE", y="RV%", hue="SOURCE_YEAR", marker="o", ci=None)

# Tweak appearance
g.set_axis_labels("Age", "RV%")
g.set_titles("Variant: {col_name}")
g.add_legend(title="Dataset Year")
g.fig.suptitle("RV% vs Age per NP300 Variant (2014–2023)", y=1.03)
g.tight_layout()
plt.show()
```

HiRider shows consistent performance across years, with smoother depreciation.

SE and Base show more volatility between years.

Other variants may reflect low volume or market-specific trends (e.g., one-off dips or unusually high RVs).
